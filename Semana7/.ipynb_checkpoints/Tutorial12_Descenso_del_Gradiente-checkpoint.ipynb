
# Tutorial 12 — **Descenso del Gradiente**: implementación desde cero (batch, mini‑batch y SGD)

 
**Objetivo**: Implementar **descenso del gradiente** (batch, mini‑batch y estocástico/SGD) para **regresión lineal** y **logística** desde cero, comprender el rol de la **tasa de aprendizaje** y visualizar la **convergencia** del costo.



<a id="material"></a>
## 1) Material de estudio



<a id="idea"></a>
### 1.1 Idea general y notación

Buscamos **mínimos** de una **función de costo** \($J(\theta$)\) moviendo los parámetros \($\theta$\) en la dirección opuesta al **gradiente**:  
\[$\theta \leftarrow \theta - \eta\,\nabla J(\theta)$\]  
donde **\($\eta$\)** es la **tasa de aprendizaje**.



<a id="costo"></a>
### 1.2 Funciones de costo y gradientes (casos usados aquí)

- **Regresión lineal (MSE)** con hipótesis \($\hat y = X\theta$\):  
  \($J(\theta)=\tfrac{1}{2m}\sum (\hat y - y)^2$\).  
  **Gradiente**: \($\nabla J(\theta)=\tfrac{1}{m} X^T(X\theta - y)$\).



- **Regresión logística** con hipótesis \($\hat y = \sigma(X\theta)$\), \($\sigma(z)=1/(1+e^{-z})$\):  
  \($J(\theta) = -\tfrac{1}{m}\sum \big[y\log\hat y + (1-y)\log(1-\hat y)\big]$\).  
  **Gradiente**: \($\nabla J(\theta)=\tfrac{1}{m} X^T(\hat y - y)$\).



<a id="variantes"></a>
### 1.3 Variantes de descenso del gradiente
- **Batch GD**: usa **todo** el conjunto para cada actualización (suave pero más costoso).
- **Mini‑batch GD**: usa lotes de tamaño \(b\) (compromiso entre estabilidad y velocidad).
- **SGD (estocástico)**: actualiza con **una sola muestra** a la vez (muy ruidoso, a veces útil).



<a id="lr"></a>
### 1.4 Tasa de aprendizaje, escalado y convergencia

- **\($\eta$\)** muy grande ⇒ puede **divergir**.  
- **\($\eta$\)** muy pequeña ⇒ **lento**.  
- **Escalar** *features* acelera y estabiliza.  
- **Criterios de paro**: nº de iteraciones, tolerancia en \($\lVert\nabla J\rVert$\) o en \($\Delta J$\), **early stopping**.



<a id="guiadas"></a>
## 2) Implementación


Funciones Auxiliares


```python

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

def add_intercept(X):
    X = np.asarray(X)
    if X.ndim == 1:
        X = X.reshape(-1, 1)
    ones = np.ones((X.shape[0], 1))
    return np.hstack([ones, X])

def standardize(X):
    X = np.asarray(X, dtype=float)
    mu = X.mean(axis=0)
    sd = X.std(axis=0, ddof=1)
    sd[sd==0] = 1.0
    return (X - mu) / sd, mu, sd

def mse_cost(X, y, theta):
    m = len(y)
    yhat = X @ theta
    return (0.5/m) * np.sum((yhat - y)**2)

def logistic_cost(X, y, theta):
    m = len(y)
    z = X @ theta
    yhat = 1.0/(1.0 + np.exp(-z))
    eps = 1e-12
    return (-1.0/m) * np.sum(y*np.log(yhat+eps) + (1-y)*np.log(1-yhat+eps))

def gradient_descent(X, y, theta0=None, lr=0.1, n_iter=1000, verbose=True):
    X = np.asarray(X, dtype=float)
    y = np.asarray(y, dtype=float).ravel()
    m, n = X.shape
    if theta0 is None:
        theta = np.zeros(n)
    else:
        theta = theta0.astype(float).copy()

    history = {"J": [], "theta": []}
    for it in range(1, n_iter+1):
        grad = (1.0/m) * (X.T @ (X @ theta - y))
        theta -= lr * grad
        J = mse_cost(X, y, theta)
        history["J"].append(J)
        if it <= 3 or it % max(1, n_iter//5) == 0:
            if verbose:
                print(f"Iteración {it:4d} | Costo (MSE) = {J:.6f} | ||grad|| = {np.linalg.norm(grad):.6f}")
        history["theta"].append(theta.copy())
    return theta, history

def minibatch_gd(X, y, theta0=None, lr=0.05, n_epochs=50, batch_size=32, shuffle=True, verbose=True):
    X = np.asarray(X, dtype=float)
    y = np.asarray(y, dtype=float).ravel()
    m, n = X.shape
    if theta0 is None:
        theta = np.zeros(n)
    else:
        theta = theta0.astype(float).copy()

    history = {"J": []}
    idx = np.arange(m)
    for ep in range(1, n_epochs+1):
        if shuffle:
            np.random.shuffle(idx)
        Xs, ys = X[idx], y[idx]
        for s in range(0, m, batch_size):
            Xe = Xs[s:s+batch_size]
            ye = ys[s:s+batch_size]
            grad = (1.0/len(ye)) * (Xe.T @ (Xe @ theta - ye))
            theta -= lr * grad
        J = mse_cost(X, y, theta)
        history["J"].append(J)
        if verbose:
            if ep <= 3 or ep % max(1, n_epochs//5) == 0:
                print(f"Época {ep:3d} | Costo (MSE) = {J:.6f}")
    return theta, history

def sgd(X, y, theta0=None, lr=0.01, n_epochs=20, shuffle=True, verbose=True):
    X = np.asarray(X, dtype=float)
    y = np.asarray(y, dtype=float).ravel()
    m, n = X.shape
    if theta0 is None:
        theta = np.zeros(n)
    else:
        theta = theta0.astype(float).copy()

    history = {"J": []}
    idx = np.arange(m)
    for ep in range(1, n_epochs+1):
        if shuffle:
            np.random.shuffle(idx)
        for i in idx:
            xi = X[i:i+1]
            yi = y[i:i+1]
            grad = (xi.T @ (xi @ theta - yi)).ravel()
            theta -= lr * grad
        J = mse_cost(X, y, theta)
        history["J"].append(J)
        if verbose:
            if ep <= 3 or ep % max(1, n_epochs//5) == 0:
                print(f"[SGD] Época {ep:3d} | Costo (MSE) = {J:.6f}")
    return theta, history

def gd_logistic(X, y, theta0=None, lr=0.1, n_iter=1000, verbose=True):
    X = np.asarray(X, dtype=float)
    y = np.asarray(y, dtype=float).ravel()
    m, n = X.shape
    if theta0 is None:
        theta = np.zeros(n)
    else:
        theta = theta0.astype(float).copy()

    history = {"J": []}
    for it in range(1, n_iter+1):
        z = X @ theta
        yhat = 1.0/(1.0 + np.exp(-z))
        grad = (1.0/m) * (X.T @ (yhat - y))
        theta -= lr * grad
        J = logistic_cost(X, y, theta)
        history["J"].append(J)
        if it <= 3 or it % max(1, n_iter//5) == 0:
            if verbose:
                print(f"Iter {it:4d} | LogLoss = {J:.6f} | ||grad|| = {np.linalg.norm(grad):.6f}")
    return theta, history
```


<a id="lin1d"></a>
### 2.1 Regresión lineal 1D — **GD batch** con MSE

Simulamos datos de una recta con ruido y entrenamos \($\theta=[b, w]$\) en forma **vectorizada**. Visualizamos la **curva de costo** y el **ajuste**.


```python

# Datos 1D (recta + ruido)
rng = np.random.default_rng(0)
x = np.linspace(-3, 3, 120)
y = 2.5*x - 1.7 + rng.normal(0, 0.9, size=len(x))

X = add_intercept(x)     # [1, x]
y = y.astype(float)

# Entrenamiento
print("Entrenando GD batch (regresión lineal 1D)...")
theta, hist = gradient_descent(X, y, lr=0.1, n_iter=300, verbose=True)
print("Parámetros aprendidos (b, w):", np.round(theta, 3))

# Curva de costo
plt.figure(figsize=(6.0, 3.8))
plt.plot(hist["J"])
plt.title("Regresión 1D — Costo (MSE) vs iteración")
plt.xlabel("Iteración"); plt.ylabel("Costo MSE")
plt.grid(True, alpha=0.3)
plt.show()

# Ajuste
yhat = X @ theta
plt.figure(figsize=(6.0, 3.8))
plt.scatter(x, y, alpha=0.7, edgecolor="none", label="Datos")
plt.plot(x, yhat, label="Recta aprendida")
plt.title("Regresión 1D — Ajuste con GD")
plt.xlabel("x"); plt.ylabel("y")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```


<a id="linmd"></a>
### 2.2 Regresión lineal **multi‑D** — **batch vs mini‑batch vs SGD**

Comparamos **convergencia** sobre un conjunto sintético. Antes **estandarizamos** *features* para acelerar el entrenamiento.


```python

# Datos multi-D
rng = np.random.default_rng(1)
m = 800
X_raw = np.column_stack([rng.uniform(-3, 3, m),
                         rng.normal(0, 1, m),
                         rng.uniform(-1, 1, m)])
true_theta = np.array([1.2, -2.0, 0.7])
y = X_raw @ true_theta + 0.8*rng.normal(0, 1, m)

# Escalado + intercepto
X_std, mu, sd = standardize(X_raw)
X = add_intercept(X_std)

print("Formas:", X.shape, "| Estandarización aplicada.")

# Batch GD
theta_b, h_b = gradient_descent(X, y, lr=0.15, n_iter=200, verbose=False)
print("Batch GD -> último costo:", h_b["J"][-1])

# Mini-batch
theta_mb, h_mb = minibatch_gd(X, y, lr=0.08, n_epochs=50, batch_size=64, verbose=False)
print("Mini-batch -> último costo:", h_mb["J"][-1])

# SGD
theta_s, h_s = sgd(X, y, lr=0.01, n_epochs=80, verbose=False)
print("SGD -> último costo:", h_s["J"][-1])
```

```python
# Curvas (una figura por gráfico)
plt.figure(figsize=(6.2, 3.8))
plt.plot(h_b["J"], label="Batch (iteraciones)")
plt.title("Costo MSE — Batch GD")
plt.xlabel("Iteración"); plt.ylabel("Costo")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(6.2, 3.8))
plt.plot(h_mb["J"], label="Mini-batch (épocas)")
plt.title("Costo MSE — Mini-batch GD")
plt.xlabel("Época"); plt.ylabel("Costo")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

```python

plt.figure(figsize=(6.2, 3.8))
plt.plot(h_s["J"], label="SGD (épocas)")
plt.title("Costo MSE — SGD")
plt.xlabel("Época"); plt.ylabel("Costo")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```


<a id="logreg"></a>
### 2.3 **Regresión logística** 2D — GD batch y **frontera de decisión**

Generamos un problema 2D linealmente separable y entrenamos con **log‑loss**. Dibujamos la **frontera** \($\sigma(X\theta)=0.5$\).


```python

# Datos 2D para clasificación
rng = np.random.default_rng(2)
m = 300
x1 = rng.normal(-1.0, 0.9, m//2)
x2 = rng.normal( 1.2, 0.9, m//2)
y  = np.r_[np.zeros(m//2), np.ones(m//2)]
X_raw = np.r_[np.c_[x1, rng.normal(0, 1, m//2)],
              np.c_[x2, rng.normal(0, 1, m//2)]]

# Estandarizar y agregar intercepto
X_std, mu, sd = standardize(X_raw)
X = add_intercept(X_std)

print("Entrenando Regresión Logística con GD...")
theta, hist = gd_logistic(X, y, lr=0.2, n_iter=300, verbose=True)
print("Theta aprendido:", np.round(theta, 3))
```

```python

# Curva de log-loss
plt.figure(figsize=(6.0, 3.8))
plt.plot(hist["J"])
plt.title("Logística 2D — LogLoss vs iteración")
plt.xlabel("Iteración"); plt.ylabel("LogLoss")
plt.grid(True, alpha=0.3)
plt.show()

```

```python
# Frontera de decisión
xx1 = np.linspace(X_std[:,0].min()-0.5, X_std[:,0].max()+0.5, 200)
xx2 = np.linspace(X_std[:,1].min()-0.5, X_std[:,1].max()+0.5, 200)
XX1, XX2 = np.meshgrid(xx1, xx2)
grid = add_intercept(np.c_[XX1.ravel(), XX2.ravel()])
Z = 1.0/(1.0 + np.exp(-(grid @ theta)))
Z = Z.reshape(XX1.shape)

plt.figure(figsize=(6.2, 4.2))
plt.contourf(XX1, XX2, Z>0.5, alpha=0.25)
plt.scatter(X_std[y==0,0], X_std[y==0,1], alpha=0.7, edgecolor="none", label="Clase 0")
plt.scatter(X_std[y==1,0], X_std[y==1,1], alpha=0.7, edgecolor="none", label="Clase 1")
plt.title("Regresión logística — Frontera 0.5 (GD)")
plt.xlabel("x1 (std)"); plt.ylabel("x2 (std)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
