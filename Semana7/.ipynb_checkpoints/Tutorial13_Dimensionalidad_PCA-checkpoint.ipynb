## 1) Material de estudio



<a id="por-que"></a>
### 1.1 ¿Por qué reducir la dimensionalidad?
- **Visualización**: facilita mirar patrones con 2–3 dimensiones.
- **Ruido/redundancia**: varias columnas aportan información repetida; reducir preserva lo esencial.
- **Eficiencia**: menos atributos ⇒ entrenamiento y cómputo más ágiles.
- **Curse of dimensionality**: en alta dimensión, distancias dejan de ser informativas.



<a id="pca-step"></a>
### 1.2 PCA — paso a paso 

**Idea**: encontrar nuevas direcciones (componentes) ortogonales que **maximizan la varianza**.  
**Pasos**:
1. **Estandarizar** (recomendado): centrar y escalar.
2. Calcular **SVD** de la matriz centrada: `Xc = U S Vᵀ`.
3. **Componentes** = columnas de `V`; **varianza explicada** = `S²/(m−1)`.
4. Proyectar: `Z = Xc · V_k`.




```python

import numpy as np

def estandarizar(X):
    X = np.asarray(X, dtype=float)
    mu = X.mean(axis=0)
    sd = X.std(axis=0, ddof=1)
    sd[sd == 0] = 1.0
    Xc = (X - mu) / sd
    return Xc, mu, sd

def pca_manual(X, k=None, verbose=True):
    """PCA con SVD (sin sklearn). Devuelve proyección Z, componentes, varianza explicada y proporciones."""
    Xc, mu, sd = estandarizar(X)
    m = Xc.shape[0]
    # SVD compacto
    U, S, Vt = np.linalg.svd(Xc, full_matrices=False)
    # Componentes principales (columnas)
    V = Vt.T
    # Varianzas por componente y proporciones
    var = (S**2) / (m - 1)
    ratio = var / var.sum()
    # Elegir k
    if k is None:
        k = V.shape[1]
    Vk = V[:, :k]
    Z = Xc @ Vk
    if verbose:
        print("PCA (manual) — resumen:")
        print("  Observaciones:", m, "| Atributos:", Xc.shape[1])
        print("  Varianza explicada (primeros 5):", np.round(ratio[:5], 4))
        print("  Varianza acumulada (k=2/3):",
              round(ratio[:2].sum(), 4), "/", round(ratio[:3].sum(), 4))
    return Z, Vk, var, ratio
```


<a id="pca-skl"></a>
### 1.3 PCA en **scikit‑learn** y selección de *k*

- `PCA(n_components=k)` para fijar *k* explícito.  
- `PCA(n_components=0.95)` para elegir automáticamente los **mínimos componentes** que expliquen ≥95% de varianza.  
- Úsalo dentro de **`Pipeline`** junto con **`StandardScaler`**.



<a id="otras"></a>
### 1.4 Otras técnicas de reducción
- **KernelPCA**: PCA no lineal mediante kernels (ej.: `rbf`).
- **TruncatedSVD**: SVD truncado (útil en datos dispersos/texto).
- **t‑SNE**: proyección no lineal para visualización 2D/3D (preserva vecindad local; no es transformador lineal).

## 2) Ejemplos



<a id="wine"></a>
### 2.1 *Workflow* con `wine`: estandarizar → PCA → visualizar

Usaremos el dataset **Wine** (13 atributos, 3 clases). Veremos **Scree plot**, **varianza acumulada** y un **scatter** en 2 componentes.


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine

data = load_wine()
X, y = data.data, data.target
feature_names = data.feature_names
print("Datos wine — forma:", X.shape, "| clases:", np.unique(y))
```

```python
# PCA manual para visualizar
Z, Vk, var, ratio = pca_manual(X, k=2, verbose=True)

# Scree plot (proporción por componente)
plt.figure(figsize=(6.0, 3.8))
plt.plot(np.arange(1, len(ratio)+1), ratio, marker="o")
plt.title("Scree plot — Proporción de varianza por componente")
plt.xlabel("Componente principal"); plt.ylabel("Proporción de varianza")
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Varianza acumulada
cum = np.cumsum(ratio)
plt.figure(figsize=(6.0, 3.8))
plt.plot(np.arange(1, len(cum)+1), cum, marker="o")
plt.axhline(0.95)  # línea de referencia (95%)
plt.title("Varianza explicada acumulada")
plt.xlabel("Número de componentes"); plt.ylabel("Acumulado")
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Dispersión en PC1 vs PC2
plt.figure(figsize=(6.2, 4.2))
for clase in np.unique(y):
    m = y == clase
    plt.scatter(Z[m, 0], Z[m, 1], alpha=0.8, label=f"Clase {clase}")
plt.title("Wine — Proyección PCA (PC1 vs PC2)")
plt.xlabel("PC1"); plt.ylabel("PC2")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```


<a id="clf"></a>
### 2.2 PCA para clasificación (pipeline)

Comparamos **`LogisticRegression`** con y sin PCA (2 componentes) usando una partición *train/test*.


```python

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, f1_score

Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)

# Sin PCA
pipe_base = Pipeline([("scaler", StandardScaler()),
                      ("clf", LogisticRegression(max_iter=2000))])
pipe_base.fit(Xtr, ytr)
yp_base = pipe_base.predict(Xte)
acc_base = accuracy_score(yte, yp_base)
f1_base = f1_score(yte, yp_base, average="macro")
print("Sin PCA -> Accuracy:", round(acc_base, 3), "| F1-macro:", round(f1_base, 3))
```

```python

# Con PCA (2 componentes)
from sklearn.decomposition import PCA
pipe_pca2 = Pipeline([("scaler", StandardScaler()),
                      ("pca", PCA(n_components=2)),
                      ("clf", LogisticRegression(max_iter=2000))])
pipe_pca2.fit(Xtr, ytr)
yp_pca = pipe_pca2.predict(Xte)
acc_pca = accuracy_score(yte, yp_pca)
f1_pca = f1_score(yte, yp_pca, average="macro")
print("Con PCA (2) -> Accuracy:", round(acc_pca, 3), "| F1-macro:", round(f1_pca, 3))
```

```python

# PCA con varianza objetivo (>=95%)
pipe_pca95 = Pipeline([("scaler", StandardScaler()),
                       ("pca", PCA(n_components=0.95)),
                       ("clf", LogisticRegression(max_iter=2000))])
pipe_pca95.fit(Xtr, ytr)
yp_pca95 = pipe_pca95.predict(Xte)
acc_pca95 = accuracy_score(yte, yp_pca95)
f1_pca95 = f1_score(yte, yp_pca95, average="macro")
print("Con PCA (>=95% var) -> Accuracy:", round(acc_pca95, 3), "| F1-macro:", round(f1_pca95, 3))
```


<a id="compare"></a>
### 2.3 Comparativa rápida: **KernelPCA**, **TruncatedSVD** y **t‑SNE** (visualización 2D)

> La meta aquí es **visualizar**; no todos son equivalentes a PCA ni sirven para lo mismo.


```python

import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import KernelPCA, TruncatedSVD
from sklearn.manifold import TSNE

# Estandarizar para métodos basados en distancia
Xstd = StandardScaler().fit_transform(X)

# KernelPCA (rbf)
kpca = KernelPCA(n_components=2, kernel="rbf", gamma=None, random_state=0)
X_kpca = kpca.fit_transform(Xstd)

plt.figure(figsize=(6.2, 4.2))
for c in np.unique(y):
    m = y == c
    plt.scatter(X_kpca[m,0], X_kpca[m,1], alpha=0.8, label=f"Clase {c}")
plt.title("KernelPCA (rbf) — 2D")
plt.xlabel("Comp 1"); plt.ylabel("Comp 2")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python


# TruncatedSVD (similar a PCA, útil en sparse/TF‑IDF)
svd = TruncatedSVD(n_components=2, random_state=0)
X_svd = svd.fit_transform(Xstd)

plt.figure(figsize=(6.2, 4.2))
for c in np.unique(y):
    m = y == c
    plt.scatter(X_svd[m,0], X_svd[m,1], alpha=0.8, label=f"Clase {c}")
plt.title("TruncatedSVD — 2D")
plt.xlabel("Comp 1"); plt.ylabel("Comp 2")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# t-SNE (visualización; no lineal)
tsne = TSNE(n_components=2, init="pca", learning_rate="auto", perplexity=30, random_state=0)

X_tsne = tsne.fit_transform(Xstd)

plt.figure(figsize=(6.2, 4.2))
for c in np.unique(y):
    m = y == c
    plt.scatter(X_tsne[m,0], X_tsne[m,1], alpha=0.8, label=f"Clase {c}")
plt.title("t‑SNE — 2D (visualización)")
plt.xlabel("Dim 1"); plt.ylabel("Dim 2")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
