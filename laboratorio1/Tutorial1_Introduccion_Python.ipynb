### 2) Estructuras básicas: Listas

Las **listas** almacenan elementos en orden y permiten elementos duplicados. Puedes acceder por índice, modificar, agregar o eliminar.

```python
# Creación y operaciones básicas con listas
notas = [25, 30, 35, 40]

print("Lista inicial:", notas)
print("Elemento en la posición 0:", notas[0])
print("Longitud de la lista (len):", len(notas))
print("Sublista (desde índice 1 hasta 2, exclusivo):", notas[1:3])
print("Último elemento: ", notas[-1])
print()

# Modificaciones
notas.append(15)
print("Después de append(15):", notas)

print()
notas[0] = 29
print("Después de modificar el primer elemento a 29:", notas)

print()
notas.remove(40)
print("Después de remove(40):", notas)

# Recorrido con for
print("Recorriendo elementos:")
for elemento in notas:
    print(" - Elemento:", elemento)

print()
# Comprensión de listas: transformar cada elemento
dobles = [n * 2 for n in notas]
print("Lista transformada (dobles):", dobles)
```

### 3) Estructuras básicas: Tuplas

Las **tuplas** son como listas **inmutables** (no se pueden modificar). Son útiles para agrupar datos que no deberían cambiar.

```python
coordenada = (33.45, -70.66)  # (latitud, longitud)
print("Tupla coordenada:", coordenada)
print("Latitud (coordenada[0]):", coordenada[0])
print("Longitud (coordenada[1]):", coordenada[1])

# Desempaquetado
lat, lon = coordenada
print("Desempaquetado: lat:", lat, "| lon:", lon)
```

### 4) Estructuras básicas: Diccionarios

Un **diccionario** almacena pares **clave-valor**. Es ideal cuando las claves tienen significado (por ejemplo, `"edad"`, `"nombre"`).

```python
persona = {"nombre": "Francisca", "edad": 20}
print("Diccionario inicial:", persona)
print()

print("Obtener valor con clave 'nombre':", persona["nombre"])
print("Obtener con .get('cargo', 'No definido'):", persona.get("cargo", "No definido"))
print()

# Agregar o actualizar
persona["cargo"] = "Profesora"
print("Después de agregar 'cargo':", persona)
print()

persona["edad"] = 30
print("Después de actualizar 'edad' a 30:", persona)
print()

# Eliminar
del persona["edad"]
print("Después de eliminar 'edad':", persona)
print()


print("Claves:", list(persona.keys()))
print("Valores:", list(persona.values()))
print("Ítems (pares clave-valor):", list(persona.items()))
```

### 5) Control de flujo: condicionales y ciclos

```python
# Condicionales: if / elif / else
edad = 15

if edad <= 5:
    etapa = "Infante"
elif edad < 18:
    etapa = "Adolescente"


print("Edad:", edad, "| Etapa:", etapa)
print()

# Ciclo for (repetición conocida)
print("For: imprimir números del 0 al 4")
for i in range(5):
    print(" i =", i)

print()
# Ciclo while (repetición controlada por condición)
print("While: contador mientras sea < 3")
contador = 0
while contador < 3:
    print(" Contador:", contador)
    contador += 1

print()

# break y continue
print("Ejemplo de break (se detiene al encontrar 5):")
for i in range(10):
    if i == 5:
        print("  Encontré 5, corto el ciclo con break.")
        break
    print("  i =", i)

print()
print("Ejemplo de continue (salta el 5):")
for i in range(3,8):
    if i == 5:
        print("  Saltando el 5 con continue.")
        continue
    print("  i =", i)
```

### 6) Funciones y buenas prácticas

```python
def calcular_promedio(numeros):
    """
    Primero se coloca una pequeña descripción de la función
    Calcula el promedio de una colección de números.

    Después los parametros y su tipo.

    Final, que retorna:
    El promedio (float).
    """
    if len(numeros) == 0:
        raise ValueError("No puedo calcular el promedio de una secuencia vacía.")
    return sum(numeros) / len(numeros)

datos = [10, 20, 30, 40]
prom = calcular_promedio(datos)
print("Datos:", datos)
print("Promedio calculado con calcular_promedio:", prom)
```

### 7) Manejo de excepciones

Las **excepciones** ayudan a manejar errores sin que el programa se detenga abruptamente.

```python
valores = ["12", "3.5", "N/A", "7"]
convertidos = []

for v in valores:
    try:
        numero = float(v)
        convertidos.append(numero)
        print(f"Convertido con éxito: {v} -> {numero}")
    except ValueError:
        print(f"Advertencia: '{v}' no es un número válido. Lo omito.")
    else:
        # Se ejecuta sólo si no hubo excepción
        print(f"Paso {v}")
    finally:
        # Siempre se ejecuta
        print() #Salto de línea

print("Lista final de valores convertidos:", convertidos)
```
